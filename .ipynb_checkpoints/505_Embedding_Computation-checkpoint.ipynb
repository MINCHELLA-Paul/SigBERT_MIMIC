````python
import torch
from transformers import AutoTokenizer, AutoModel

MODEL_DISTIL = "distilbert-base-uncased"
model_name = "emilyalsentzer/Bio_ClinicalBERT"
````

````python
# 1. Load the tokenizer
# This handles converting your text into numbers (tokens)
tokenizer = AutoTokenizer.from_pretrained(model_name)

# 2. Load the model
# This loads the actual weights of DistilBERT
model = AutoModel.from_pretrained(model_name)
````

````python
device = "cpu"
model = model.to(device)
````

````python
import os
import pandas as pd

# 1. Configuration
input_folder = "data"
output_folder = "arora_results_processed"
N_files = 1  # Or the specific number of files you wish to process
os.makedirs(output_folder, exist_ok=True)

# 2. Processing Loop
for k in range(1, N_files + 1):
    # Construct the file path dynamically
    file_path = os.path.join(input_folder, f"final_arora_part_{k}.csv")
    
    # Check if the file exists before processing
    if not os.path.exists(file_path):
        print(f"File {file_path} not found. Skipping...")
        continue
        
    print(f"--- Processing File {k}/{N_files}: {file_path} ---")
    
    # Load the specific chunk
    df_chunked = pd.read_csv(file_path)
    
    # Apply the Arora Methods (SIF Embeddings)
    # The variables (tokenizer, model, device) must be pre-defined in your notebook
    df_arora = arora_methods(
        df_chunked,
        tokenizer, 
        model, 
        device,
        var_cible="TEXT_CLEAN", # Ensure this column exists in your CSV
        no_cls=True,
        a_sifembedding = 1e-3
    )

    # 3. Clean up the output to save space
    # Dropping heavy word embeddings and original text as requested
    df_output = df_arora.drop(columns=['TEXT_CLEAN', 'word_embeddings'], errors='ignore')
    
    # 4. Export the specific result
    output_filename = f"arora_results_part_{k}.csv"
    output_path = os.path.join(output_folder, output_filename)
    df_output.to_csv(output_path, index=False, encoding='utf-8-sig')
    
    print(f"---------------  Saved: {output_path} !!! ^^ ---------------")

print("\n--- Pipeline Execution Completed ---")
````
